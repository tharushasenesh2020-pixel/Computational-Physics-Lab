```python
import numpy as np
import matplotlib.pyplot as plt

fs = 1000
t = np.linspace(0, 1, fs, endpoint=False)

f1, A1 = 100, 2
f2, A2 = 200, 1
sin1 = A1*np.sin(2*np.pi*f1*t)
sin2 = A2*np.sin(2*np.pi*f2*t)

result = sin1 + sin2
fft_result = np.fft.fft(result)
freq = np.fft.fftfreq(len(fft_result), 1/fs)

positive_mask = freq>0
positive_freq = freq[positive_mask]
positive_magnitudes = np.abs(fft_result)[positive_mask]* 2/fs

plt.plot(t[:100], result[:100])
plt.title("Resultant wave in Time domain")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude (V)")
plt.grid(True)
plt.show()

plt.plot(positive_freq, positive_magnitudes)
plt.title('Frequency Components (FFT)')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Amplitude (V)')
plt.grid(True)
plt.show()

max_magnitudes = positive_magnitudes[positive_magnitudes > 0.1]
max_freq = positive_freq[positive_magnitudes > 0.1]
print("Detected Frequencies and Amplitudes:")
for freq, amp in zip(max_freq, max_magnitudes):
    print(f"Frequency: {freq:.1f} Hz, Amplitude: {amp:.2f} V")
```
