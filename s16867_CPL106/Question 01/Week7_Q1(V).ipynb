```python
import numpy as np
import matplotlib.pyplot as plt

def sinewave(t, A, F, phi):
    return A*np.sin(2*np.pi*F*t+phi)

A = 0.1
F = 50
phi = np.pi/2

t_start = 0
t_end = 4 * (1 / F)  

sampling_frequencies = [10, 100, 1000, 10000] 

t_continuous = np.linspace(t_start, t_end, 500) 
y_continuous = sinewave(t_continuous, A, F, phi)
plt.plot(t_continuous, y_continuous, label='Original Signal (50 Hz)', color='black',)

for fs in sampling_frequencies:
    t_sampled = np.arange(t_start, t_end, 1 / fs)
    y_sampled = sinewave(t_sampled, A, F, phi)
    plt.plot(t_sampled, y_sampled, label=f'Sampled at {fs} Hz', linestyle='--')

plt.title('Sinewave Sampled at Different Frequencies')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude (m)')
plt.legend()
plt.grid(True)
plt.show()
```
