```python
import numpy as np
def runge_kutta_second_order(f, x0, t0, dt, num_step):

    t = [t0]
    x = [np.array(x0)]

    t_1 = t0
    x_1 = np.array(x0)

    for _ in range(num_step):
        k_1 = dt*f(t_1, x_1)
        k_2 = dt*f(t_1 + dt,x_1 + k_1)
        t_2 = t_1 + dt

        x_2 = x_1 + 0.5 * (k_1 + k_2)
        t_2 = t_1 + dt

        t.append(t_2)
        x.append(x_2)

        t_1 = t_2
        x_1 = x_2
        
    return t,x
def voltage(t, x):
    v_in = 5
    rc = 2
    return (v_in-x)/rc

x0 = 0
t0 = 0
total_time = 4
num_step = 500

dt = total_time/ num_step
time_values, v_out = runge_kutta_second_order(voltage, x0, t0, dt, num_step)

final_voltage_RK = v_out[-1]
final_time_RK = time_values[-1]

print(f"Analytical solution for x(4) = {4.6875:.4f}")
print(f"RK method second order result at t = {final_time_RK:.2f} hours: {final_voltage_RK:.4f} V")
```
