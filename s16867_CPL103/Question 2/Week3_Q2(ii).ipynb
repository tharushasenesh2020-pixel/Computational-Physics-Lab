```python
def euler_method(f, x0, t0, dt, num_step):

    t = [t0]
    x = [x0]

    t_1 = t0
    x_1 = x0

    for _ in range(num_step):
        dx_dt = f(t_1, x_1)
        x_2 = x_1 + dt*(dx_dt)
        t_2 = t_1 + dt

        t.append(t_2)
        x.append(x_2)

        t_1 = t_2
        x_1 = x_2
        
    return t,x
```
