```python
def euler_method(f, x0, t0, dt, num_step):

    t = [t0]
    x = [x0]

    t_1 = t0
    x_1 = x0

    for _ in range(num_step):
        dx_dt = f(t_1, x_1)
        x_2 = x_1 + dt*(dx_dt)
        t_2 = t_1 + dt

        t.append(t_2)
        x.append(x_2)

        t_1 = t_2
        x_1 = x_2
        
    return t,x

def voltage(t, x):
    v_in = 5
    rc = 2
    return (v_in-x)/rc

x0 = 0
t0 = 0
total_time = 4
num_step = 500

dt = total_time/ num_step
time_values, v_out = euler_method(voltage, x0, t0, dt, num_step)

final_voltage_euler = v_out[-1]
final_time_euler = time_values[-1]

print(f"Analytical solution for x(4) = {4.6875:.4f}")
print(f"Euler's Method result at t = {final_time_euler:.2f} hours: {final_voltage_euler:.4f} V")
```
