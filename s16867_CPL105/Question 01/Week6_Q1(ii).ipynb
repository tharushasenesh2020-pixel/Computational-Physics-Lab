```python
def bisection(f, a, b, tol):
    if f(a) * f(b) >= 0:
       print("The function must have opposite signs at a and b.")

    while (b - a) / 2 > tol:
        c = (a + b) / 2
        if f(c) == 0:
            return c

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c
            
    return (a + b) / 2
```
