```python
import numpy as np
import matplotlib.pyplot as plt

n = 100                
sa = 1                 
sb = 1                
ns = 100         

xa = np.random.rand(n) * 50        
ya = np.random.rand(n) * 50         
xb = np.random.rand(n) * 50 + 50   
yb = np.random.rand(n) * 50        

tcal = np.zeros(ns)  
tcar = np.zeros(ns)  
tcbl = np.zeros(ns)  
tcbr = np.zeros(ns) 

for step in range(ns):
    cal = 0  
    car = 0  
    cbl = 0 
    cbr = 0 

    for i in range(n):
        if 0 <= xa[i] <= 50:
            cal += 1
        else:
            car += 1
        if 0 <= xb[i] <= 50:
            cbl += 1
        else:
            cbr += 1

        r = np.random.rand()
        if r < 0.5 and xa[i] > sa:  
            xa[i] -= sa
        elif r >= 0.5 and xa[i] < 100 - sa: 
            xa[i] += sa

        r = np.random.rand()
        if r < 0.5 and xb[i] > sb:  
            xb[i] -= sb
        elif r >= 0.5 and xb[i] < 100 - sb:  
            xb[i] += sb

    tcal[step] = cal / n
    tcar[step] = car / n
    tcbl[step] = cbl / n
    tcbr[step] = cbr / n

plt.figure(figsize=(10, 6))
time_steps = range(ns)
plt.plot(time_steps, tcal, label='A in Left (0-50)', color='red')
plt.plot(time_steps, tcar, label='A in Right (50-100)', color='red', linestyle='--')
plt.plot(time_steps, tcbl, label='B in Left (0-50)', color='green')
plt.plot(time_steps, tcbr, label='B in Right (50-100)', color='green', linestyle='--')
plt.xlabel('Time Steps')
plt.ylabel('Concentration')
plt.title('Diffusion of Particles A and B Over Time')
plt.legend()
plt.grid(True)
plt.show()
```
