```python
import random
import numpy as np

ladders = {5: 14, 10: 28, 32: 45}
snakes = {15: 7, 37: 19, 49: 34}

def simulate_game():
    position = 1
    rolls = 0

    while position < 50:
        roll = random.randint(1, 6)
        rolls += 1

        if position + roll <= 50:
            position += roll

            if position in ladders:
                position = ladders[position]
            elif position in snakes:
                position = snakes[position]
    return rolls

def monte_carlo_simulation(num_games=10000):
    roll_counts = [simulate_game() for _ in range(num_games)]

    expected_rolls = np.mean(roll_counts)
    prob_within_25 = np.mean(np.array(roll_counts) <= 25)

    return expected_rolls, prob_within_25

if __name__ == "__main__":
    num_games = 10000
    expected_rolls, prob_within_25 = monte_carlo_simulation(num_games)

    print(f"Simulated {num_games} games")
    print(f"Expected number of rolls to finish: {expected_rolls:.2f}")
    print(f"Probability of finishing within 25 rolls: {prob_within_25:.4f}")
```
