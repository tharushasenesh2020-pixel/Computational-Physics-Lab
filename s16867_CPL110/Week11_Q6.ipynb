```python
import numpy as np
import matplotlib.pyplot as plt

nominal_resistance = 470
tolerance = 0.05
num_resistors = 1000
voltage = 12

resistance_values = np.random.uniform(nominal_resistance * (1 - tolerance), nominal_resistance * (1 + tolerance), num_resistors)
current_values = voltage / resistance_values

plt.figure(figsize=(10, 6))
plt.hist(resistance_values, bins=30, edgecolor='black', color='blue')
plt.title('Distribution of Resistance Values')
plt.xlabel('Resistance')
plt.ylabel('Frequency')
plt.show()

plt.figure(figsize=(10, 6))
plt.hist(current_values, bins=30, edgecolor='black', color='green')
plt.title('Distribution of Current Values (V=12V)')
plt.xlabel('Current (A)')
plt.ylabel('Frequency')
plt.show()
```
