```python
import random
import matplotlib.pyplot as plt

def calculate_pi_monte_carlo(num_points):
   
    points_in_circle = 0
    x_in_circle = []
    y_in_circle = []
    x_outside_circle = []
    y_outside_circle = []

    for _ in range(num_points):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)

        distance = x**2 + y**2

        if distance <= 1:
            points_in_circle += 1
            x_in_circle.append(x)
            y_in_circle.append(y)
        else:
            x_outside_circle.append(x)
            y_outside_circle.append(y)

    pi_estimate = 4 * (points_in_circle / num_points)

    plt.figure(figsize=(8, 8))
    plt.scatter(x_in_circle, y_in_circle, color='blue', s=1, label='Inside Circle')
    plt.scatter(x_outside_circle, y_outside_circle, color='red', s=1, label='Outside Circle')
    
    square_x = [-1, 1, 1, -1, -1]
    square_y = [-1, -1, 1, 1, -1]
    plt.plot(square_x, square_y, color='black', linestyle='--')
    
    circle = plt.Circle((0, 0), 1, color='green', fill=False, linestyle='-')
    plt.gca().add_patch(circle)

    plt.title(f'Monte Carlo Pi Approximation (N = {num_points})\nEstimated Pi = {pi_estimate:.5f}')
    plt.xlabel('X-axis')
    plt.ylabel('Y-axis')
    plt.gca().set_aspect('equal', adjustable='box')
    plt.legend()
    plt.grid(True)
    plt.show()

    return pi_estimate

if __name__ == '__main__':
    points_to_test = [100, 1000, 10000, 100000, 1000000]
    
    for n in points_to_test:
        pi_value = calculate_pi_monte_carlo(n)
        difference = abs(pi_value - 3.141592653589793)
        print(f" {"Estimated Pi":<20} | {"Difference from True Pi":<25}")
        print(f" {pi_value:<20.6f} | {difference:<25.6f}")
```
