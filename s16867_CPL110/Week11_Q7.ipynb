```python
import numpy as np
import matplotlib.pyplot as plt

V_in = 10
R1_nominal = 10000
R2_nominal = 2000

tolerance_R1 = 0.05
tolerance_R2 = 0.02

num_simulations = 100000

R1_min = R1_nominal * (1 - tolerance_R1)
R1_max = R1_nominal * (1 + tolerance_R1)
R2_min = R2_nominal * (1 - tolerance_R2)
R2_max = R2_nominal * (1 + tolerance_R2)

R1_values = np.random.uniform(R1_min, R1_max, num_simulations)
R2_values = np.random.uniform(R2_min, R2_max, num_simulations)

V_out_values = V_in * (R2_values / (R1_values + R2_values))

plt.figure(figsize=(10, 6))
plt.hist(V_out_values, bins=50, edgecolor='black', alpha=0.7)
plt.title(f'Monte Carlo Simulation of Voltage Divider Output (N={num_simulations})')
plt.xlabel('Output Voltage ($V_{out}$)')
plt.ylabel('Frequency')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axvline(np.mean(V_out_values), color='red', linestyle='dashed', linewidth=2, label=f'Mean V_out: {np.mean(V_out_values):.4f} V')
plt.legend()
plt.tight_layout()
plt.show()
```
