```python
import numpy as np
import matplotlib.pyplot as plt

x = np.random.uniform(low=-1, high=1, size = 2000) 

plt.hist(x, bins=50, density=True, edgecolor='black', alpha=0.7)
plt.title('Histogram of Uniformly Distributed Random Numbers from -1 to +1')
plt.xlabel('Value')
plt.ylabel('Density')
plt.grid(axis='y', alpha=0.75)

plt.show()
```
