```python
import tkinter as tk
from tkinter import ttk, messagebox

def button_click():
    try:
        if direction_var.get() == "CtoFK":
            if not entry0.get():
                messagebox.showerror("Error", "Please enter a Celsius value")
                return
            temp = float(entry0.get())
            if temp < -273.15:
                messagebox.showerror("Error", "Temperature cannot be below absolute zero (-273.15°C)")
                return
            temp_F = temp * (9/5) + 32
            temp_K = temp + 273.15
            entry1.delete(0, tk.END)
            entry2.delete(0, tk.END)
            entry1.insert(0, f"{temp_F:.2f}")
            entry2.insert(0, f"{temp_K:.2f}")
        else:
            if not entry1.get():
                messagebox.showerror("Error", "Please enter a Fahrenheit value")
                return
            temp_F = float(entry1.get())
            if temp_F < -459.67:
                messagebox.showerror("Error", "Temperature cannot be below absolute zero (-459.67°F)")
                return
            temp_C = (temp_F - 32) * (5/9)
            temp_K = temp_C + 273.15
            entry0.delete(0, tk.END)
            entry2.delete(0, tk.END)
            entry0.insert(0, f"{temp_C:.2f}")
            entry2.insert(0, f"{temp_K:.2f}")
    except ValueError:
        messagebox.showerror("Error", "Please enter a valid number")

def clear():
    entry0.delete(0, tk.END)
    entry1.delete(0, tk.END)
    entry2.delete(0, tk.END)

# Create the main application window
root = tk.Tk()
root.title("Temperature Converter")
root.geometry("350x200")
root.resizable(False, False)

# Configure style
style = ttk.Style()
style.configure("TLabel", font=("Helvetica", 12))
style.configure("TEntry", font=("Helvetica", 12))
style.configure("Convert.TButton", font=("Helvetica", 10), padding=5, background="#4CAF50", foreground="black")
style.map("Convert.TButton", background=[('active', '#45a049')])  
style.configure("Clear.TButton", font=("Helvetica", 10), padding=5, background="#f44336", foreground="black")
style.map("Clear.TButton", background=[('active', '#d32f2f')])  

# Create label widgets
label0 = ttk.Label(root, text="Celsius (°C):")
label1 = ttk.Label(root, text="Fahrenheit (°F):")
label2 = ttk.Label(root, text="Kelvin (K):")
label0.grid(row=0, column=0, padx=5, pady=5, sticky=tk.W)
label1.grid(row=1, column=0, padx=5, pady=5, sticky=tk.W)
label2.grid(row=2, column=0, padx=5, pady=5, sticky=tk.W)

# Create entry widgets
entry0 = ttk.Entry(root, width=10)
entry0.grid(row=0, column=1, padx=5, pady=5)
entry1 = ttk.Entry(root, width=10)
entry1.grid(row=1, column=1, padx=5, pady=5)
entry2 = ttk.Entry(root, width=10)
entry2.grid(row=2, column=1, padx=5, pady=5)

# Create conversion direction radio buttons
direction_var = tk.StringVar(value="CtoFK")
radio_c_to_fk = ttk.Radiobutton(root, text="C to F/K", variable=direction_var, value="CtoFK")
radio_f_to_ck = ttk.Radiobutton(root, text="F to C/K", variable=direction_var, value="FtoCK")
radio_c_to_fk.grid(row=3, column=0, padx=5, pady=5, sticky=tk.W)
radio_f_to_ck.grid(row=3, column=1, padx=5, pady=5, sticky=tk.W)

# Create button widgets
button1 = ttk.Button(root, text="Convert", command=button_click, style="Convert.TButton")
button1.grid(row=0, column=2, padx=5, pady=5)
button2 = ttk.Button(root, text="Clear", command=clear, style="Clear.TButton")
button2.grid(row=1, column=2, padx=5, pady=5)

# Start the main event loop
root.mainloop()
```
