```python
import numpy as np
import matplotlib.pyplot as plt

r = 0.5
n = 200
x0 = 0.1
x_val = np.zeros(n) 
x_val[0] = x0

for i in range(n-1): 
    x_val[i+1] = r * x_val[i] * (1 - x_val[i]) 

plt.plot(x_val)
plt.title("Xn vs n")
plt.xlabel("n")
plt.ylabel("Xn")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

r = 0.5
n = 200
initial_val = [0.1, 0.3, 0.5, 0.7, 0.9]

plt.figure(figsize=(12, 8))

for i, x0 in enumerate(initial_val):
    x_val = np.zeros(n)
    x_val[0] = x0
    
    for j in range(1, n):
        x_val[j] = r * x_val[j-1] * (1 - x_val[j-1])
    
    plt.subplot(3, 2, i + 1)
    plt.plot(range(n), x_val, label=f'$x_0$ = {x0}')
    plt.title("Xn vs n")
    plt.xlabel("n")
    plt.ylabel("$x_n$")
    plt.legend()
    plt.grid()

plt.tight_layout()
plt.show()
```
