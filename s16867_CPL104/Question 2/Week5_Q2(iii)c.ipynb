```python
import numpy as np

alpha = np.pi/3
beta = np.pi/4
delta = np.pi/4
gamma = 3*np.pi/2

def exp(theta):
    return np.cos(theta)+1j*np.sin(theta)

A1 = np.array([[exp(-beta/2), 0],
               [0, exp(beta/2)]])

A2 = np.array([[np.cos(gamma/2), -np.sin(gamma/2)],
               [np.sin(gamma/2), np.cos(gamma/2)]])

A3 = np.array([[exp(-delta/2), 0],
               [0, exp(delta/2)]])

global_phase = exp(alpha)
U = global_phase*A1@A2@A3

print("Matrix U is: ", U)

```
