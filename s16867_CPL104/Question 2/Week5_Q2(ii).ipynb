# (a)

```python
import numpy as np
def matrix_sum(A, B):

    if A.shape != B.shape:
        print("Error: Matrices must have the same dimensions for addition.")
    return A+B

A = np.array([[5,2], [2,4]])
B = np.array([[9,6], [3,8]])

Sum = matrix_sum(A, B)
print("A+B:\n", Sum)
    
```

# (b)

```python
import numpy as np
def det_of_matrix(matrix):

    if matrix.shape[0] != matrix.shape[1]:
        print("Must be a square matrix for calculate determinant")
    return np.linalg.det(matrix)

A = np.array([[5, 2], [1, 6]])

det = det_of_matrix(A)
print("Matrix A:\n", A)
print("Determinant of A: ",det)
```

# (c)

```python
import numpy as np

def transpose_matrix(A):
    return A.T

A = np.array([[3, 4, 1],
              [1, 7, 4]])
transpose = transpose_matrix(A)
print("Transpose of A: \n",transpose)
```

# (d)

```python
import numpy as np

def check_symmetric(A):

    if A.shape[0] != A.shape[1]:
        return False
    return np.array_equal(A, A.T)

A = np.array([[4, 2, 1],
              [2, 8, 5],
              [1, 5, 2]])

print ("Matrix A: \n", A)
print ("Is a symmetric?", check_symmetric(A))
```

# (e)

```python
import numpy as np

def matrix_multiply(A, B):

    if A.shape[1] != B.shape[0]:
        print ("Number of columns must be equal")
    return np.dot(A, B)

A = np.array([[3,2],
              [6,1]])
B = np.array([[2,9],
              [6,2]])

A_dot_B =  matrix_multiply(A, B)
print ("A.B: \n", A_dot_B)
    
```
